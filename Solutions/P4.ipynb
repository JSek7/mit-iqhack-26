```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# 1) Load the circuit
path = r"C:\Users\joven\Documents\MIT Iqhack 26\Files\P4_gentle_mound.qasm"
qc = QuantumCircuit.from_qasm_file(path)

# 2) Add measurements (MPS works via sampling)
qc_m = qc.copy()
qc_m.measure_all()

# 3) Create MPS simulator
sim = AerSimulator(method="matrix_product_state")

# 4) Set MPS accuracy / speed tradeoff
sim.set_options(
    matrix_product_state_max_bond_dimension=128,
    matrix_product_state_truncation_threshold=1e-12,
)

# 5) Run sampling
shots = 50_000
result = sim.run(
    qc_m,
    shots=shots,
    seed_simulator=1
).result()

# 6) Extract dominant bitstring
counts = result.get_counts()
dominant = max(counts, key=counts.get)

print("Dominant bitstring:", dominant)
print("Estimated probability:", counts[dominant] / shots)
```
